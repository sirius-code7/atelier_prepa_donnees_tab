### 0. Importer les bibliothèques nécessaires 

**Déf :** 

-`NumPy est une bibliothèque open-source du langage Python conçue pour le calcul scientifique et numérique. Elle permet de manipuler facilement des tableaux et des matrices de données de grande taille (en une ou plusieurs dimensions), un peu comme une super-calculatrice mathématique ultra-rapide.`

-`Pandas est une bibliothèque open-source du langage Python conçue pour la manipulation et l'analyse de données. Elle permet de travailler facilement avec des données tabulaires (lignes et colonnes), un peu comme un tableau Excel ou une table SQL`

-`Matplotlib est une bibliothèque open-source du langage Python conçue pour la visualisation de données. Elle permet de créer facilement des graphiques clairs et personnalisés (courbes, histogrammes, nuages de points), un peu comme un crayon et une feuille de dessin pour illustrer vos chiffres.`

-`Scikit-learn est une bibliothèque open-source du langage Python conçue pour l'apprentissage automatique (machine learning). Elle permet d'entraîner facilement des modèles pour faire des prédictions ou classer des données, un peu comme une boîte à outils intelligente capable d'apprendre à partir d'exemples passés`

### 1.1 : Chargement des données

La fonction `pd.read_csv()` permet de charger un fichier CSV dans un DataFrame Pandas.

Le fichier `smart_building_raw.csv` contient les données brutes provenant des capteurs des bâtiments intelligents.

Les données sont importées dans un DataFrame nommé `df`, qui sera utilisé pour toute la suite de l'atelier.

```python
import pandas as pd

df = pd.read_csv("../data/smart_building_raw.csv")

df
```

### 1.2 : Afficher les premières lignes

La méthode `head()` permet d'afficher les premières lignes d'un DataFrame.

Par défaut, elle affiche les 5 premières lignes.

Cette commande permet d'avoir un premier aperçu de la structure et du contenu des données.

```python
df.head()
```

### 1.3 : Afficher les dernières lignes

La méthode `tail()` permet d'afficher les dernières lignes d'un DataFrame.

Par défaut, elle affiche les 5 dernières lignes.

Cette observation permet de vérifier que le dataset est correctement chargé jusqu'à sa dernière observation.

```python
df.tail()
```

### 1.4 : Nombre d'observations

La propriété `shape` permet d'obtenir les dimensions d'un DataFrame.

Le premier élément correspond au nombre de lignes.

Chaque ligne représente une observation, c'est-à-dire une mesure enregistrée par les capteurs.

```python
df.shape[0]
```

### 1.5 : Nombre de variables

Le deuxième élément de la propriété `shape` correspond au nombre de colonnes.

Chaque colonne représente une variable du dataset.

```python
df.shape[1]
```

### 1.6 : Afficher les noms des variables

La propriété `columns`, combinée à `select_dtypes()`, permet d'afficher les noms de toutes les colonnes numériques du DataFrame.


```python
df.select_dtypes(include="number").columns
```

#### Identifier les variables numériques

Les variables numériques sont les colonnes contenant des valeurs quantitatives.

À l'aide de la fonction `select_dtypes()`, les variables numériques suivantes ont été identifiées :

- `id_mesure`
- `temperature`
- `humidite`
- `co2`
- `occupation`
- `consommation_kwh`

La colonne `id_mesure` est numérique, mais elle correspond à un identifiant et ne sera pas considérée comme une variable quantitative pour les analyses statistiques et les corrélations.

### 1.7 : Noms des variables catégorielles

La propriété `.columns`, combinée à `select_dtypes()`, permet d'obtenir uniquement les noms des colonnes de type textuel (chaînes de caractères) du DataFrame.

```python
df.select_dtypes(include="object").columns
```

#### Identifier les variables catégorielles

Les variables catégorielles sont les colonnes contenant des valeurs représentant des catégories ou des groupes.

Les variables catégorielles identifiées sont :

- `batiment`
- `type_batiment`
- `zone`
- `mode_climatisation`
- `etat_systeme`
- `jour_semaine`
- `alerte`

La colonne `date` apparaît également parmi les variables de type `object`, mais elle semble correspondre à une variable temporelle stockée sous forme de texte. Elle sera vérifiée et convertie dans la question suivante.
