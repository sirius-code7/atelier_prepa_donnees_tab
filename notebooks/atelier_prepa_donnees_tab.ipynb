### 10: Déterminer les statistiques descriptives

Les statistiques descriptives permettent de résumer les principales caractéristiques des variables numériques.

Les indicateurs demandés sont :

- la moyenne ;
- la médiane ;
- le minimum ;
- le maximum ;
- l'écart-type ;
- les quartiles.

La variable `id_mesure` est exclue, car elle correspond à un identifiant et non à une mesure quantitative.

```python
variables_numeriques = [
    "temperature",
    "humidite",
    "co2",
    "occupation",
    "consommation_kwh"
]

statistiques = pd.DataFrame({
    "moyenne": df[variables_numeriques].mean(),
    "mediane": df[variables_numeriques].median(),
    "minimum": df[variables_numeriques].min(),
    "maximum": df[variables_numeriques].max(),
    "ecart_type": df[variables_numeriques].std(),
    "q1": df[variables_numeriques].quantile(0.25),
    "q2": df[variables_numeriques].quantile(0.50),
    "q3": df[variables_numeriques].quantile(0.75)
})

statistiques
```

#### notes sur les stats descriptives

- La moyenne représente la valeur moyenne des observations. = somme/ nbr
- La médiane représente la valeur centrale lorsque les données sont classées. = milieu
- Le minimum représente la plus petite valeur observée.
- Le maximum représente la plus grande valeur observée.
- L'écart-type mesure la dispersion des données autour de la moyenne.
- Q1 correspond au premier quartile : 25 % des valeurs lui sont inférieures ou égales.
- Q2 correspond à la médiane.
- Q3 correspond au troisième quartile : 75 % des valeurs lui sont inférieures ou égales.

### 12.a : Rechercher les humidités inférieures à 0

Une humidité exprimée en pourcentage ne peut pas être négative.

Nous recherchons donc les observations dont la valeur de `humidite` est inférieure à `0`.


```python
df[df["humidite"] < 0]
```

### 12.b : Rechercher les humidités supérieures à 100

Une humidité exprimée en pourcentage doit être comprise entre `0` et `100`.

Nous recherchons donc les observations dont la valeur de `humidite` est supérieure à `100`.


```python
df[df["humidite"] > 100]
```

### 12.c : Rechercher les températures extrêmement élevées

Une température très élevée peut indiquer une erreur de mesure.

Le sujet ne fournit pas de seuil précis. Pour effectuer un premier contrôle, nous considérons ici qu'une température supérieure à `50 °C` mérite une vérification.


```python
df[df["temperature"] > 50]
```

### 12.d : Rechercher les valeurs négatives

Certaines variables numériques ne peuvent pas contenir de valeurs négatives.

Nous vérifions les valeurs inférieures à `0` pour les variables numériques pertinentes afin d'identifier les éventuelles incohérences.


```python
variables_a_verifier = [
    "temperature",
    "humidite",
    "co2",
    "occupation",
    "consommation_kwh"
]

for colonne in variables_a_verifier:
    valeurs_negatives = df[df[colonne] < 0]

    print(f"\n{colonne} : {len(valeurs_negatives)} valeur(s) négative(s)")
    display(valeurs_negatives)
```

### 12.e : Rechercher les consommations négatives

Une consommation énergétique ne peut pas être négative.

Nous recherchons donc les observations dont `consommation_kwh` est inférieure à `0`.
