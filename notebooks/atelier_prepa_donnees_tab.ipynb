### 10: Déterminer les statistiques descriptives

Les statistiques descriptives permettent de résumer les principales caractéristiques des variables numériques.

Les indicateurs demandés sont :

- la moyenne ;
- la médiane ;
- le minimum ;
- le maximum ;
- l'écart-type ;
- les quartiles.

La variable `id_mesure` est exclue, car elle correspond à un identifiant et non à une mesure quantitative.

```python
variables_numeriques = [
    "temperature",
    "humidite",
    "co2",
    "occupation",
    "consommation_kwh"
]

statistiques = pd.DataFrame({
    "moyenne": df[variables_numeriques].mean(),
    "mediane": df[variables_numeriques].median(),
    "minimum": df[variables_numeriques].min(),
    "maximum": df[variables_numeriques].max(),
    "ecart_type": df[variables_numeriques].std(),
    "q1": df[variables_numeriques].quantile(0.25),
    "q2": df[variables_numeriques].quantile(0.50),
    "q3": df[variables_numeriques].quantile(0.75)
})

statistiques
```

#### notes sur les stats descriptives

- La moyenne représente la valeur moyenne des observations. = somme/ nbr
- La médiane représente la valeur centrale lorsque les données sont classées. = milieu
- Le minimum représente la plus petite valeur observée.
- Le maximum représente la plus grande valeur observée.
- L'écart-type mesure la dispersion des données autour de la moyenne.
- Q1 correspond au premier quartile : 25 % des valeurs lui sont inférieures ou égales.
- Q2 correspond à la médiane.
- Q3 correspond au troisième quartile : 75 % des valeurs lui sont inférieures ou égales.

### 12.f : Transformer les valeurs manifestement erronées en valeurs manquantes

Lorsqu'une valeur est manifestement incorrecte et qu'il est impossible de retrouver sa véritable valeur, elle est transformée en valeur manquante.

Nous utilisons `NaN` afin de conserver l'information indiquant qu'une donnée est absente, sans inventer une nouvelle valeur.


```python
# Humidité : une valeur doit être comprise entre 0 et 100 %
df.loc[
    (df["humidite"] < 0) | (df["humidite"] > 100),
    "humidite"
] = np.nan

# Température : seuil de contrôle retenu précédemment
df.loc[
    df["temperature"] > 50,
    "temperature"
] = np.nan

# CO2 : une concentration ne peut pas être négative
df.loc[
    df["co2"] < 0,
    "co2"
] = np.nan

# Occupation : un nombre de personnes ne peut pas être négatif
df.loc[
    df["occupation"] < 0,
    "occupation"
] = np.nan

# Consommation énergétique : elle ne peut pas être négative
df.loc[
    df["consommation_kwh"] < 0,
    "consommation_kwh"
] = np.nan
```

### 13.c : Quelle stratégie utiliser pour les valeurs manquantes ?

La stratégie dépend du type de variable et de la quantité de valeurs manquantes.

Pour cet atelier :

* les variables numériques pourront être imputées avec la médiane ;
* les variables catégorielles pourront être imputées avec la valeur la plus fréquente ;
* la suppression des lignes ne sera utilisée que si elle est réellement justifiée.

La médiane sera privilégiée pour les variables numériques car elle est moins sensible aux valeurs extrêmes que la moyenne.


### 13.d : Peut-on supprimer toutes les lignes contenant des valeurs manquantes ?

Supprimer toutes les lignes contenant au moins une valeur manquante peut entraîner une perte importante d'observations.

Cette stratégie n'est donc pas retenue automatiquement. Il est préférable d'analyser la quantité de données manquantes et d'utiliser une stratégie d'imputation lorsque cela est pertinent.


```python
lignes_supprimees = df.isna().any(axis=1).sum()

print("Lignes contenant au moins une valeur manquante :", lignes_supprimees)
print("Nombre total de lignes :", len(df))
```

### 13.e : Dans quels cas utiliser la moyenne ?

La moyenne peut être utilisée pour remplacer des valeurs manquantes d'une variable numérique lorsque les données sont relativement homogènes et peu influencées par des valeurs extrêmes.

Elle est moins adaptée lorsque la distribution est fortement asymétrique ou contient des valeurs aberrantes considérables.


### 13.f : Quand préférer la médiane ?

La médiane est préférable lorsque les données sont asymétriques ou contiennent des valeurs extrêmes.

Elle est moins influencée par ces valeurs que la moyenne, ce qui en fait une stratégie adaptée aux variables numériques de ce dataset.


### 13.g : Comment traiter une variable catégorielle ?

Pour une variable catégorielle, une stratégie possible consiste à remplacer les valeurs manquantes par la catégorie la plus fréquente.

Cette stratégie correspond à une imputation par le mode.

### 15 : Identifier visuellement les distributions, la dispersion et les valeurs extrêmes

Nous allons analyser visuellement les variables `temperature`, `humidite`, `co2` et `consommation_kwh`.

Les histogrammes permettent d'observer la forme des distributions et leur éventuelle symétrie.

Les boxplots permettent d'observer la dispersion, les valeurs extrêmes et les valeurs potentiellement aberrantes.

La variable `id_mesure` est exclue car elle correspond à un identifiant.


```python
variables_analyse = [
    "temperature",
    "humidite",
    "co2",
    "consommation_kwh"
]

for colonne in variables_analyse:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    sns.histplot(
        data=df,
        x=colonne,
        kde=True,
        ax=axes[0]
    )

    axes[0].set_title(f"Distribution de {colonne}")
    axes[0].set_xlabel(colonne)
    axes[0].set_ylabel("Fréquence")

    sns.boxplot(
        data=df,
        x=colonne,
        ax=axes[1]
    )

    axes[1].set_title(f"Boxplot de {colonne}")
    axes[1].set_xlabel(colonne)

    plt.tight_layout()
    plt.show()
```
